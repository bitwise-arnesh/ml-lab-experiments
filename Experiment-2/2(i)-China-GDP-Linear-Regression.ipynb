# ML Lab Experiment 2(i)

## China GDP - Linear Regression

### Objective

Split the given China GDP dataset into training and testing data in a 70:30 ratio. Perform Linear Regression using the training data and evaluate the model using the testing data with Mean Squared Error (MSE).

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Load the China GDP dataset
df = pd.read_csv("LR_or_PR_china_gdp.csv")

# Separate the input feature and target variable
# Year is the input and Value is the target
X = df[["Year"]]
y = df["Value"]

# Split the data into 70% training and 30% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)

# Create the Linear Regression model
model = LinearRegression()

# Train the model using only the training data
model.fit(X_train, y_train)

# Display the slope and intercept
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Predict the values for the training data
y_train_pred = model.predict(X_train)

# Predict the values for the testing data
y_test_pred = model.predict(X_test)

# Calculate Mean Squared Error for training data
train_mse = mean_squared_error(y_train, y_train_pred)

# Calculate Mean Squared Error for testing data
test_mse = mean_squared_error(y_test, y_test_pred)

print("Training MSE:", train_mse)
print("Testing MSE:", test_mse)
```

```python
# Plot training data and regression line
plt.scatter(X_train, y_train, label="Training Data")
plt.plot(X_train, y_train_pred, label="Linear Regression Line")

plt.xlabel("Year")
plt.ylabel("GDP Value")
plt.title("China GDP - Training Data")
plt.legend()
plt.show()
```

```python
# Plot testing data and regression line
plt.scatter(X_test, y_test, label="Testing Data")
plt.plot(X_test, y_test_pred, label="Linear Regression Line")

plt.xlabel("Year")
plt.ylabel("GDP Value")
plt.title("China GDP - Testing Data")
plt.legend()
plt.show()
```

## Conclusion

The China GDP dataset was divided into training and testing data in a 70:30 ratio. A Linear Regression model was trained using the training data and evaluated using the testing data with Mean Squared Error (MSE).

The graphs show the actual training and testing data along with the Linear Regression line. The data shows a nonlinear growth pattern, so the straight regression line does not closely follow all the data points.
