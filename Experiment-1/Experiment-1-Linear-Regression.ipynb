# ML Lab Experiment 1
## Linear Regression

### Objective

Perform Linear Regression using the given training data
and evaluate the model using the given testing data using MSE.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

# Load training and testing data
train_data = pd.read_csv("train_data - train_data.csv")
test_data = pd.read_csv("test_data - test_data.csv")

# Separate input and target variables
X_train = train_data[["x"]]
y_train = train_data["y"]

X_test = test_data[["x"]]
y_test = test_data["y"]

# Create Linear Regression model
model = LinearRegression()

# Train the model using training data
model.fit(X_train, y_train)

# Predict using testing data
y_pred = model.predict(X_test)

# Calculate Mean Squared Error
mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error:", mse)
```

```python
# Predict training data
y_train_pred = model.predict(X_train)

# Plot training data
plt.scatter(X_train, y_train, label="Training Data")

# Plot regression line
plt.plot(X_train, y_train_pred, label="Linear Regression Line")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression - Training Data")
plt.legend()
plt.show()
```

```python
# Predict testing data
y_test_pred = model.predict(X_test)

# Plot testing data
plt.scatter(X_test, y_test, label="Testing Data")

# Plot regression line
plt.plot(X_test, y_test_pred, label="Linear Regression Line")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression - Testing Data")
plt.legend()
plt.show()
```

```python
# Calculate MSE for training and testing data
train_mse = mean_squared_error(y_train, y_train_pred)
test_mse = mean_squared_error(y_test, y_test_pred)

print("Training MSE:", train_mse)
print("Testing MSE:", test_mse)
```

### Conclusion

The Linear Regression model was trained using the given training dataset and evaluated using the given testing dataset. The regression line captured the overall positive relationship between x and y in both datasets.

The Mean Squared Error (MSE) on the testing data was approximately 352.1068, indicating the prediction error of the model on unseen data.
